# RTS-GMLC dataset overview

Read-only exploration of the checksummed canonical datasets.
No training, no final-test access beyond descriptive statistics of the
published artifacts — the research pipeline lives in `scripts/`.

```python
from pathlib import Path
import pyarrow.parquet as pq

ROOT = Path.cwd().parent
for target in ("load", "wind", "pv"):
    t = pq.read_table(ROOT / "data" / "processed" / f"{target}_hourly.parquet")
    df = t.to_pylist()
    col = [v for k, v in df[0].items() if k != "timestamp"][0]
    vals = [float(r[list(r.keys())[1]]) for r in df]
    print(f"{target}: {len(df)} rows, {t.num_columns} columns, first={vals[0]:.2f}, last={vals[-1]:.2f}")
```
